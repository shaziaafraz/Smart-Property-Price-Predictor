```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_dataset.csv")
df.head()
```

```python
df['Price_in_Lakhs'].hist(bins=30)
plt.title("Distribution of Price")
plt.show()

df['Size_in_SqFt'].hist(bins=30)
plt.title("Distribution of Size")
plt.show()
```

```python
city_cols = [col for col in df.columns if col.startswith("City_")]

# melt the city one-hot columns into one column
df_city = df[city_cols + ['Price_per_SqFt']].copy()
df_city = df_city.melt(id_vars='Price_per_SqFt', var_name='City', value_name='flag')
df_city = df_city[df_city['flag'] == 1]

sns.boxplot(x='City', y='Price_per_SqFt', data=df_city)
plt.xticks(rotation=90)
plt.show()
```
